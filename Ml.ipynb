```python
import numpy as np
import pandas as pd

from sklearn.linear_model import LinearRegression
```

```python
#  single linear regression model

# house prediction dataset
X = np.array([[1], [2], [3], [4], [5]]) # 2 dimensional array  / area
y = np.array([2, 4, 6, 8, 10]) # 1 dimensional array / price

model = LinearRegression()
model.fit(X, y) # independent variable and dependent variable

prediction = model.predict(np.array([[6]])) # predict the price of house with area 6

print("Predicted price of house with area 6:", prediction[0])
```

```python
import matplotlib.pyplot as plt

x = np.array([1, 2, 3, 4, 5])
y = np.array([-2, 4, 6, 6, 10])

plt.plot(x,y, marker="o", color="red",markersize=10,linewidth=5,markerfacecolor="blue",markeredgecolor="green")
plt.xlabel("Area")
plt.ylabel("Price")
plt.title("Housing Data")

plt.savefig("plot.png") # save the plot as a png file
plt.show()
```

```python
import numpy as np
from sklearn.neighbors import KNeighborsClassifier

x = np.array([[6,2], [4,3], [6,5], [7,8], [5,5]])
y = np.array([0, 0, 1, 1, 1]) # 0 = fail, 1 = pass

model = KNeighborsClassifier(n_neighbors=3)
model.fit(x, y)

prediction = model.predict(np.array([[6, 4]])) # predict the result of student with marks 5 and 4

print("Predicted result of student with marks 6 and 4:", prediction[0])
```

```python
import pandas as pd
from sklearn.neighbors import KNeighborsClassifier

df = pd.read_csv('dataclass.csv') # read the dataset from csv file

df_x = df[['math', 'cs']] # independent variables
df_y = df['rate'] # dependent variable

model = KNeighborsClassifier(n_neighbors=3)
model.fit(df_x, df_y)

data_predict = [[6, 4]] # data to predict

prediction = model.predict(data_predict) # predict the result of student with marks 6 and 4
print("Predicted result of student with marks 6 and 4:", prediction[0])
```
